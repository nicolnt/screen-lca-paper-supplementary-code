## Prepare the subset dataframe

```python
indicators = ['PRODQNT', 'EXPQNT', 'IMPQNT']
```

```python
# NOTE: Relevant NACE codes:
# - 20.14 Other organic basic chemicals (en)
# - 20.16 Plastics in primary forms (en)
# Source : 05-02-2026 - https://showvoc.op.europa.eu/#/datasets/ESTAT_Statistical_Classification_of_Economic_Activities_in_the_European_Community_Rev._2.1._(NACE_2.1)/
NACE_codes_filter = ['2014', '2016'] 
NACE_codes_filter
```

```python
# NOTE: Filter the DataFrame on
# - 'freq' = 'A'
# - 'reporter' = 'EU27_2020'
# - 'product' code starts with either '2014' or '2016'
# - 'indicators' = 'PRODQNT', 'EXPQNT' or 'IMPQNT'

annual_series = new_df.loc[
    (new_df.index.get_level_values('freq') == 'A')
    & (new_df.index.get_level_values('reporter') == 'EU27_2020')
    & new_df.index.get_level_values('product').str.startswith(tuple(NACE_codes_filter))
    # & new_df.index.get_level_values('product').str.startswith(tuple(cpa_codes_filter.to_list())) # CPA codes obtained from EXIOBASE matching
    & new_df.index.get_level_values('indicators').isin(indicators)
    # & (new_df.index.get_level_values('indicators').str.endswith('QNT') | new_df.index.get_level_values('indicators').str.endswith('VAL'))
]
annual_series
```

```python
unit_rows = new_df.loc[
    (new_df.index.get_level_values('freq') == 'A')
    & (new_df.index.get_level_values('reporter') == 'EU27_2020')
    & new_df.index.get_level_values('product').str.startswith(tuple(NACE_codes_filter))
    & (new_df.index.get_level_values('indicators') == 'QNTUNIT')
]
unit_rows
```

## Forecast

```python
products = annual_series.index.get_level_values('product').unique()
products
```

```python
# NOTE: Estimate quantity at a certain date
# Source : 29-01-2026 - https://codepointtech.com/how-to-find-first-row-that-meets-criteria-in-pandas-effectively/

def forecast(row_series):

    row_name = row_series.name
    
    criteria = (row_series.str.strip() != ':') & row_series.str.isnumeric()

    # NOTE: Take the lastest available data index
    last_match_index = criteria.nlargest(1, keep='last').index

    matching_row = row_series.loc[last_match_index]
    value = int(matching_row.iloc[0])
    year = last_match_index[0]

    # NOTE: Retrieve unit
    unit = unit_rows.loc[(unit_rows.index.get_level_values('product') == row_name[2])].loc(axis=1)[last_match_index].iloc[0].iloc[0]

    # NOTE: Retrive product name
    # description = ''
    # description_row = prodcom_classification.loc[lambda x : x['PRODUCT (Codes)'] == row_name[2]]
    # if len(description_row):
    #     description = description_row.loc(axis=1)['PRODUCT (Labels)'].iloc[0]

    # first_match_index = criteria.idxmax()
    # Retrieve the row using .loc
    # first_matching_row = series.loc[first_match_index]

    # return series.loc[lambda x: x.str.strip() != ':'].tail(1).astype('int').values
    
    return (value, year, unit)
```

```python
pd.DataFrame(
        annual_series.apply(forecast, axis=1, result_type='expand'),
        index=annual_series.index,
        columns=['latest value', 'latest year', 'latest unit']
    )
```

```python
latest = annual_series.apply(forecast, axis=1, result_type='expand')
latest.columns=['latest value', 'latest year', 'latest unit']

annual_series_latest = annual_series.join(latest)

# annual_series[['latest value', 'latest year', 'latest unit']] = annual_series.apply(forecast, axis=1, result_type='expand')
```

## Transform result

```python
annual_series_latest.loc[annual_series_latest.index.get_level_values('indicators').isin(['PRODQNT']), 'latest value']
```

```python
# NOTE: Generate the new MultiIndex
reorg_index = annual_series_latest.index
reorg_index = reorg_index.droplevel('indicators')
reorg_index = reorg_index.drop_duplicates()
reorg_index
```

```python
# NOTE: Create a new DataFrame using the new MultiIndex
reorg_df = pd.DataFrame(index=reorg_index)
reorg_df
```

```python
# NOTE: Get the label name of a Prodcom code
def get_label(code):
    description = ''
    description_row = prodcom_classification.loc[lambda x : x['PRODUCT (Codes)'] == code]
    if len(description_row):
        description = description_row.loc(axis=1)['PRODUCT (Labels)'].iloc[0]
    return description
```

```python
# NOTE: Insert Prodcom product label text column
reorg_df['product label'] = ''
reorg_df['product label'] = reorg_df.apply(lambda row: get_label(row.name[2]), axis=1)
reorg_df
```

```python
# NOTE: Return the value for a specific product key ('row_name') and for the selected indicator
def get_column_value(row_name, indicator, column_name):
    row = annual_series_latest.loc[
        (annual_series_latest.index.get_level_values('freq') == row_name[0])
        & (annual_series_latest.index.get_level_values('reporter') == row_name[1])
        & (annual_series_latest.index.get_level_values('product') == row_name[2])
        & (annual_series_latest.index.get_level_values('indicators') == indicator),
        column_name
    ]
    if len(row):
        return row.iloc[0]
    else:
        return 'N/A'
```

```python
# NOTE: Generate 3 columns ('value', 'year', 'unit') for each indicator ('PRODQNT', 'EXPQNT', 'IMPQNT')
for indicator in indicators:
    reorg_df[indicator + ' value'] = reorg_df.apply(lambda row: get_column_value(row.name, indicator, 'latest value'), axis=1)
    reorg_df[indicator + ' year'] = reorg_df.apply(lambda row: get_column_value(row.name, indicator, 'latest year'), axis=1)
    reorg_df[indicator + ' unit'] = reorg_df.apply(lambda row: get_column_value(row.name, indicator, 'latest unit'), axis=1)
reorg_df
```
